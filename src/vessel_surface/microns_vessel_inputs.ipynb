# Vessel inputs for the surface distances
Run the one cell below. It downloads `vessel_inputs.zip` to your computer (Downloads). Nothing else to do here.

```python
# Downloads the vessel data that only Colab can reach and hands it to you as ONE zip:
#   Wan & Wei's TriSAM vessel mesh (BossDB wei2024, ~50 MB) and the 11 Shih lab vascular skeletons
#   (How et al. 2026, ~40 MB). Nothing is computed here; the labelling and the distances are done
#   afterwards from this zip, with the same rules as the VR viewer.
import urllib.request, zipfile, os, time
SKEL = ["skelFinal_PAs_v7_connected", "skelFinal_ACTs_v7", "skelFinal_terminalPAs_v7",
        "skelFinal_AVs_v7_connected", "skelFinal_CVTs_v7", "skelFinal_terminalAVs_v7",
        "skelFinal_PCVs_v7", "skelFinal_deepCVTs_v7", "skelFinal_terminalPCVs_v7",
        "skelFinal_BBs_v7", "skelFinal_v7_branch_order"]
GET = {"trisam_mesh.bin": "https://bossdb-open-data.s3.amazonaws.com/wei2024/minnie/bv/mesh/1%3A0%3A276-1562_240-1133_247-871"}
for s in SKEL:
    GET["shih/" + s + ".skel"] = "https://storage.googleapis.com/microns_vascular_skeleton/" + s + "/skeletons/3"
    GET["shih/" + s + ".info"] = "https://storage.googleapis.com/microns_vascular_skeleton/" + s + "/skeletons/info"
out = "/content/vessel_inputs.zip"
with zipfile.ZipFile(out, "w", zipfile.ZIP_DEFLATED) as z:
    for name, url in GET.items():
        t = time.time()
        try:
            data = urllib.request.urlopen(url, timeout=300).read()
        except Exception as e:
            if name.endswith(".info"): print("  (no", name, ":", e, ")"); continue
            raise
        z.writestr(name, data); print(f"{name}: {len(data)/1e6:.1f} MB ({time.time()-t:.0f} s)")
print("zip:", round(os.path.getsize(out) / 1e6, 1), "MB")
from google.colab import files
files.download(out)
```
